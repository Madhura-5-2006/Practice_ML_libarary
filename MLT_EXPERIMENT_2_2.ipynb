
#step 2 : read the data

```python
df=pd.read_csv('/content/drive/MyDrive/Salary_dataset.csv')
```


#step 3 : visualizing the data

```python
sns.lmplot(x='YearsExperience',y='Salary',data=df)
```

#step 4 : predicting the value using linear equation y=mx+b
#storing this regression into a variable named model

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python

model.fit(df[['YearsExperience']],df[['Salary']])
```

```python

model.score(df[['YearsExperience']],df[['Salary']])
```
